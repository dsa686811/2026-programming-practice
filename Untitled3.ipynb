```python
import tensorflow as tf
from tensorflow import keras
import numpy as np

# 1. 영화 리뷰 데이터셋(IMDB) 불러오기 (단어 1만 개 기준)
vocab_size = 10000
imdb = keras.datasets.imdb
(train_data, train_labels), (test_data, test_labels) = imdb.load_data(num_words=vocab_size)

# 문장 길이 맞추기 (패딩)
train_data = keras.preprocessing.sequence.pad_sequences(train_data, value=0, padding='post', maxlen=256)
test_data = keras.preprocessing.sequence.pad_sequences(test_data, value=0, padding='post', maxlen=256)

# AI 모델 구성 및 학습
model = keras.Sequential([
    keras.layers.Embedding(vocab_size, 16),
    keras.layers.GlobalAveragePooling1D(),
    keras.layers.Dense(16, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

print("--- AI가 문장 감성 학습을 시작합니다 ---")
model.fit(train_data, train_labels, epochs=3, batch_size=512, verbose=0)

# 단어 사전 가져오기
word_index = imdb.get_word_index()
word_index = {k: (v + 3) for k, v in word_index.items()}
word_index["<PAD>"] = 0
word_index["<START>"] = 1
word_index["<UNK>"] = 2
word_index["<UNUSED>"] = 3
reverse_word_index = {value: key for (key, value) in word_index.items()}

print("✨ 1단계 완료: 감성 분석 학습이 끝났습니다!")
```

```python
def review_encode(text):
    # 특수문자 제거 및 소문자 변환
    encoded = [1] # 시작 기호
    for w in text.lower().split():
        if w in word_index:
            encoded.append(word_index[w])
        else:
            encoded.append(2) # 사전에 없는 단어는 <UNK> 처리
    return keras.preprocessing.sequence.pad_sequences([encoded], value=0, padding='post', maxlen=256)

print("✨ 2단계 완료: 문장 번역 함수 준비 완료!")
```

```python
# 테스트할 영어 문장 (원하는 문장으로 자유롭게 바꿔서 실행해보세요!)
test_sentence ="This was a complete waste of time, absolutely terrible and boring."
# 또는 부정적인 문장 예시: "This movie was terrible, boring and awful."

# 예측 실행
encode_input = review_encode(test_sentence)
prediction = model.predict(encode_input)[0][0]

print(f"입력한 문장: 「 {test_sentence} 」")
print(f"AI 확률 점수: {prediction:.4f}")

if prediction > 0.5:
    print("✨ [AI 판정 결과]: 😃 매우 긍정적(Positive)인 내용입니다!")
else:
    print("✨ [AI 판정 결과]: 😡 매우 부정적(Negative)인 내용입니다!")
```
