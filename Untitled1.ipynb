```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt

# 1. 가벼운 공개 데이터셋(Fashion MNIST) 불러오기
fashion_mnist = keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

# 분류할 옷 이름 목록
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

# 2. 데이터 전처리 (0~255 픽셀값을 0~1 사이로 정규화)
train_images = train_images / 255.0
test_images = test_images / 255.0

# 3. 나만의 Keras 딥러닝 모델 구성하기
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),      # 28x28 2차원 이미지를 1줄로 펴기
    keras.layers.Dense(128, activation='relu'),      # 은닉층 (특징 학습)
    keras.layers.Dropout(0.2),                       # 과적합 방지
    keras.layers.Dense(10, activation='softmax')     # 출력층 (10개 카테고리 중 확률 계산)
])

# 모델 컴파일
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# 4. 모델 학습하기 (5번 반복 학습)
print("--- 모델 학습을 시작합니다 ---")
model.fit(train_images, train_labels, epochs=5)

# 5. 모델 성능 평가하기
test_loss, test_acc = model.evaluate(test_images, test_labels, verbose=2)
print(f"\n테스트 정확도: {test_acc*100:.2f}%")

# 6. 새로운 데이터로 추론(예측)해보고 결과 확인하기
predictions = model.predict(test_images[:5])

plt.figure(figsize=(12, 3))
for i in range(5):
    plt.subplot(1, 5, i + 1)
    plt.xticks([])
    plt.yticks([])
    plt.grid(False)
    plt.imshow(test_images[i], cmap=plt.cm.binary)

    predicted_label = np.argmax(predictions[i])
    true_label = test_labels[i]

    color = 'blue' if predicted_label == true_label else 'red'
    plt.xlabel(f"Pred: {class_names[predicted_label]}\n(True: {class_names[true_label]})", color=color)
plt.show()
```
